Starter notebook with short NumPy, Pandas, and SQL examples for the Week 6 writeup.

Run this notebook in a Python 3 environment with pandas and numpy installed.

```python
import numpy as np
import pandas as pd

# NumPy example
arr = np.array([1, 2, 3, 4.5])
print('arr:', arr)
print('mean:', arr.mean())

# Pandas example
df = pd.DataFrame({'city': ['Accra','Lagos','Kumasi'], 'mean_temp': [27.5, 28.3, 26.8]})
print('
DataFrame sample:')
print(df)

# SQLite example (in-memory)
import sqlite3
conn = sqlite3.connect(':memory:')
cur = conn.cursor()
cur.execute('CREATE TABLE weather(city TEXT, precipitation REAL);')
cur.executemany('INSERT INTO weather VALUES(?,?)', [('Accra', 12.4), ('Lagos', 5.2), ('Kumasi', 20.1)])
for row in cur.execute('SELECT city, SUM(precipitation) as total_precip FROM weather GROUP BY city'):
    print(row)
conn.close()
```
